# Data Preprocessing and Visualization

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Re-creating sample data for standalone use
df = pd.DataFrame({
    'person_income': [50000, 60000, 70000, 80000, 1000000],
    'loan_status': [0, 0, 1, 0, 1]
})

plt.figure(figsize=(10, 5))
sns.histplot(df['person_income'].clip(upper=df['person_income'].quantile(0.99)), kde=True)
plt.title('Capped Income Distribution')
plt.show()
```
