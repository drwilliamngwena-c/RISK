# Model Interpretations and Business Drivers
This notebook provides a deep dive into the logic behind the Credit Risk models.

## 1. Probability of Default (PD)
**Key Driver: Income**
In our Logistic Regression model, income has a negative coefficient, meaning as income increases, the log-odds of default decrease. This aligns with the business intuition that higher earners have better financial buffers.

## 2. Loss Given Default (LGD)
**Key Driver: Recovery Potential**
LGD measures the loss a bank faces if a default occurs. We model this based on the assets and recovery potential of the borrower.

## 3. Exposure at Default (EAD)
**Key Driver: Utilization**
EAD tracks how much the borrower owes at the moment of default. We use regression to estimate this based on current outstanding balances and credit limits.

### Regulatory Alignment
By using interpretable models (Logistic/Linear Regression), we ensure that the models meet Basel II/III transparency requirements, allowing for clear auditing of risk weights.
